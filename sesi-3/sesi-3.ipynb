```python
# List
# membuat lis => []
# mutable = > biasa kita ubah

nama_siswa = ["cliarine", "fadhlan", "hafiz", "anton"]
# nama_siswa = [] = list kosong
print("list nama siswa", nama_siswa)
# menggunakan method untuk menambah dan menghapus data dalam list
# method menambah => appen (), insert()

# append/menambah data di posisi akhir
nama_siswa.append("budi")
print("nama siswa baru: ", nama_siswa)

# insert/ menambah data di posisi tertentu
nama_siswa.insert(1, "joni")
print("nama siswa joni ditambahkan: ", nama_siswa)

# remove() & pop()
# remove() =>harus mengetahui value yang ingin dihapus
nama_siswa.remove("fadhlan") #menghapus data
print("upadate list nama siswa", nama_siswa)

# pop() => menghapus berdasarkan indeks
nama_siswa.pop(3)
print("update list 2:", nama_siswa)

# list indexing => mengakses elemen tertentu di dalam list
print("nama siswa berprestasi:", nama_siswa [2])

# list slicing => memotong list = mengakses beberapa value dalam list
print(nama_siswa [0:2]) #start : stop
print(nama_siswa [1:])#start:
print(nama_siswa [:2])# :end
print(nama_siswa [-1])# :end

# melihat panjang list
print("panjang list:", len(nama_siswa))

# duplicate elemen
nama_siswa.append("budi")
print("nama budi ada 2:", nama_siswa)

# dalam set boleh masukin berbagai tipe data seperti integer, float, dll

```

```python
# Set => {}
# tidak bisa menerima duplicate elemen
# tidak ada urutan

himpunan_nama = {"gio", "fadhlan", "clairine"}
print("set awal:", himpunan_nama)

# menambahkan set
himpunan_nama.add("nimas")

print("set update:", himpunan_nama)

# menghapus data dalam set
# method remove()
himpunan_nama.remove("fadhlan")
print("set update setelah di remove:", himpunan_nama)

# method discard()
himpunan_nama.discard("Nimas")
print("set update setelah di remove:", himpunan_nama)

# gabungan/join, selisih, intersection(irisan)
himpunan_siswa_a = {"clairine", "fadhlan", "gio"}
himpunan_siswa_b = {"fadhlan", "hafiz", "anton"}
print("gabungan nama siswa:", himpunan_siswa_a | himpunan_siswa_b)

# selisih => -
print("selisih himp a dan himp b: ", himpunan_siswa_a - himpunan_siswa_b)
print("selisih himp b dan himp a: ", himpunan_siswa_b - himpunan_siswa_a)

# irisan
print("irisan himp a & b", himpunan_siswa_a & himpunan_siswa_b)
```

```python
# Dictionary => kamus {key:value}

kontak_telp = {
    "Clairine": 998,
    "fadhlan": 997,
    "gio": 996
}
print("dictionary awal:", kontak_telp)
# menambahkan data ke dalam dict => indexing
kontak_telp["hafiz"] = 995
print("dictionary update:", kontak_telp)
# menrubah data
kontak_telp ["hafiz"] = 776
print("dictionary update:", kontak_telp)

# menghapus data 
kontak_telp.pop("gio")
print("dictionary update:", kontak_telp)

# del
del kontak_telp ["fadhlan"]
print("dictionary update:", kontak_telp)

# Keys(), Values(), Items()
# keys( => mengambil semua keys)
print(kontak_telp.keys())

# values() => mengambil semua value
print(kontak_telp.values())

# items( => mengambil semua key dan value)
print(kontak_telp.items())



```

```python
# Function
"""
A function in Python is a reusable block of code that performs a specific task, 
defined using the def keyword. Functions can take parameters, which are variables listed inside the parentheses of the function definition. 
These parameters act as placeholders for values (called arguments) that you provide when calling the function, allowing the function to work flexibly with different inputs.
"""

# Contoh Function
def perkenalan(nama):
    print(f"halao nama saya {nama}")

perkenalan ("clairine")
perkenalan("fadhlan")

"""
sintax function:

def nama_function (parameter 1, parameter 2 dst):
    blok isi kode yang ingin di eksekusi

#panggil nama functiobn (isi parameter)
"""
def perkenalan (nama):
    print(f"halo nama saya {nama}")

perkenalan("nia")

# jenis-jenis function
# function tanpa parameter
def sapa():
    print ("halo teman-teman semua")

# panggil function
sapa()

# function dengan parameter
def sapa(nama, jenis_kelamin):
    print(f"halo saya {nama}, jenis kelamin saya {jenis_kelamin}")

sapa("nia", "wanita")
sapa("alamsyah", "ayam")

# function dengan nilai kembali/return value
def tambah (angka1, angka2):
    return angka1 + angka2


# panggil function
hasil = tambah(10,2)
print(hasil)
print(tambah(1, 2))


# function dengan default parameter
def sapa (nama, lokasi = "indonesia"):
    print(f"halo nama saya {nama}, saya tinggal di {lokasi}")
sapa("fadhlan", "brussles")
```

```python
# Chatbot sederhana

# library random
import random
# Respon pertanyaan
respons = {
    "hai": ["halo apa kabar", "halo", " selamat datang"],
    "nama kamu siapa": ["saya chatbot", "saya jokowi", "saya guru" ],
    "bye": ["selamat tinggal", "have nice day", "bye bitch"]
}

def chatbot():
    # minta input (user input)
    # cek kondisi (inputan user = data di respons)

    #perulangan untuk cek kondisi
    while True:
        user_input = input("user:").lower()
        #cek user input apakah ada di respons
        if user_input in respons:
            print("Bot: ", random.choice(respons[user_input]))
        elif user_input == "keluar":
            print("Bot: sampai jumpa")
            break
        else: print("maaf saya tidak mengerti")

print(type(respons))

# panggil function
chatbot()
```
