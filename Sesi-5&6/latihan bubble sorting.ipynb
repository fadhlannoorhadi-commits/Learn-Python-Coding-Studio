```python
produk = [{"nama produk": "Printer", "harga": 30000},
{"nama produk": "handuk", "harga": 20000}, 
{"nama produk": "AC", "harga": 15000},
{"nama produk": "Meja", "harga": 32000},
{"nama produk": "Ransel", "harga": 400000},]

def tampilkan_produk(data):
    print("\n daftar produk")

for p in produk:
    print(f"nama produk: {p["nama produk"]}, harga produk: {p["harga"]}")


def bubble_sort_by_nama(data_list):

    pjglist = len(data_list)

    for index in range(0, pjglist - 1):

        for jindex in range(0, pjglist - index - 1):
            if data_list [jindex] ["nama produk"] > data_list [jindex + 1] ["nama produk"]:

                data_list [jindex], data_list [jindex + 1] = data_list [jindex + 1],   data_list [jindex]

    return data_list


def bubble_sort_by_harga(data_list):

    pjglist = len(data_list)

    for index in range(0, pjglist - 1):

        for jindex in range(0, pjglist - index - 1):
            if data_list [jindex] ["harga"] > data_list [jindex + 1] ["harga"]:

                data_list [jindex], data_list [jindex + 1] = data_list [jindex + 1],   data_list [jindex]

    return data_list

def main():
    print("selamat datang di toko kami")

    while True:
        print("\n1.urutkan berdasarkan nama")
        print("2. urutkan berdasarkan harga")
        print("3. tampilkan semua produk")
        print("4. terimakasih telah berkunjung")
     
        user_input = input("pilih servis 1/2/3/4:")

        if user_input == "1":
            hasil = bubble_sort_by_nama(produk)
            tampilkan_produk(hasil)
        elif user_input == "2":
            hasil = bubble_sort_by_harga(produk)
            tampilkan_produk(hasil)
        elif user_input == "3":
            tampilkan_produk(produk)
        elif user_input == "4":
            print("terimakasih puki")
            break
        else: print("maaf saya tidak mengerti coba ulangi kembali anjg")


main()
        
```
