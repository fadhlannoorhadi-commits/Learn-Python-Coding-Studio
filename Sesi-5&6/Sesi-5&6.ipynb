```python
# Sorting
"""
Cara mengurutkan suatu data yang cukup besar. 
Berdasarkan ukuran, nilai, atau urutan alfabet
"""

# Bubble sort - (Visualize)
# pseudocode: kode yang seperti bahasa manusuia
# Bubble Sort
"""
Algoritma pseudocode
(kode yang seperti bahasa manusia)
bubble sort:
listawal : [5, 3, 8, 4, 2] 
mulai
pjglist = len(list awal)
untuk index dari 0 ke pjglist-1
    untuk jindex dari 0 ke pjglist-index-1:
           if arr[jindex] > arr[jindex+1]:
                tukar arr[jindex] dan arr[jindex + 1]
selesai
"""
def bubble_sort(data_list):
# mulai
    pjglist = len(data_list) #ambil panjang list
#lakukan looping sepanjang list
    for index in range(0, pjglist-1):
    # lakukan loopong dalam list untuk cek besar kecil data
    
        for jindex in range (0, pjglist - index - 1):
        #bandingkan data yang bersebelahan
         if data_list [jindex] > data_list[jindex+1]:
            #swap posisinya
            data_list[jindex], data_list[jindex + 1] = data_list[jindex + 1], data_list[jindex]

    return data_list

#cara menggunakan function
print(bubble_sort([50,90,20,30,60]))

list_nilai = [100,69,67,34,53,84]
print("nilai awal:", list_nilai)
print("nilai terurut:", bubble_sort (list_nilai))
```

```python
# Object Oriented Programing (OOP)
# Encapsulation = menyembunyikan suatu data / atribut agar tidak bisa diakses langsung dari luar kelas
# biasanya digunakan untuk melindungi data
class AkunBank:
    def __init__(self, nama, saldo):
        self.nama = nama
        self.__saldo = saldo #private saldo atribut
        #perhatikan penggunaan __
    
    #setter and getter
    def setSaldo(self, saldo):
        self.__saldo = saldo
    def getSaldo (self):
        print(f"saldo kamu {self.__saldo}")

    # membuat object/akun
akun1 = AkunBank ("Hafiz",7000)

print(akun1.nama)

akun1.getSaldo()
akun1.setSaldo(10000000000)
akun1.getSaldo()
```

```python
# Inheritance/pewarisan
# sebuah kelas bisa mewarisi atribut dan method dari kelas lain

# parent class/super class
class Hewan: 
    def __init__(self, nama):
        self.nama = nama
    def suara(self):
        print("Hewan Bersuara")
    def jumlah_kaki(self):
        print("Hewan Berkaki 4")
#child class
class Kucing(Hewan):
    def suara(self):#kelas kucing mewarisi dari kelas hewan
        print("miaw")
class Anjing(Hewan): #kelas anjing mewarisi dari kelas hewan
    def suara(self):
        print("guk guk")

#buat objek
mochi = Kucing ("mochy")
choky = Anjing("Choky")

mochi.suara()
choky.suara()
choky.jumlah_kaki()

    
```

```python
# Polimorphisme
# objek berbeda bisa memiliki method dengan nama yang sama tapi implementasinya berbeda

class Person:
    def __init__(self, nama):
        self.nama = nama
    def greet(self):
        print("halo teman-teman")

class English(Person):
    def greet(self):
        print("Hello friends")


class German(Person):
    def greet(self):
        print("tschuss preunds")

    # buat objek
person1 = English("Hafiz")
person2 = German ("Gio")
person3 = English ("nayda")

person1.greet()
person2.greet()
person3.greet()

def greet(): #ini function
    print("halo teman teman")

greet()
    
```
