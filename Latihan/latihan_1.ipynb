```python
# Latihan Control Flow
angka = 4
if angka % 2 == 0:
    print("angka ini genap")
else: print("angka ini ganjil")
print(angka)
```

# Penjelasan Kode Ganjil Genap Diatas
angka % 2 menghitung sisa pembagian angka dengan 2
Jika hasilnya 0, berarti angka tersebut habis dibagi 2 → genap
Jika hasilnya bukan 0, berarti tidak habis dibagi 2 → ganjil
