```python
import tkinter as tk
window = tk.Tk()
window.title("etalase dengan GUI")
window.geometry ("900x800")
window.configure(bg="lightblue")


# data etalase toko jam
etalase = [{"nama": "Rolex", "harga":750000000}, 
{"nama": "Casio", "harga": 500000000}, 
{"nama": "Timex", "harga": 250000000},
{"nama": "Swatch", "harga": 100000000}, 
{"nama": "Fossil", "harga": 150000000},
{"nama": "Seiko", "harga": 300000000},
{"nama": "Citizen", "harga": 350000000},
{"nama": "Tissot", "harga": 400000000},
{"nama": "Omega", "harga": 600000000},
{"nama": "Tag Heuer", "harga": 550000000},
{"nama": "Patek Philippe", "harga": 800000000,},
{"nama": "Audemars Piguet", "harga": 900000000},
{"nama": "Hublot", "harga": 650000000},
{"nama": "Breitling", "harga": 450000000},
{"nama": "Longines", "harga": 200000000}]

# function to display etalase
def tampilkan_etalase(etalase):
    hasil_text = "Etalase Toko Jam :\n\n"
    for e in etalase:
        hasil_text += f"Nama: {e['nama']} | Harga: {e['harga']}\n"

    label_output.config(text=hasil_text)

# bubble sort functions
def bubble_sort_by_nama(data_list):
    pjglist = len(data_list)
    for index in range(0, pjglist - 1):
        for jindex in range(0, pjglist - index - 1):
            if data_list[jindex]["nama"] > data_list[jindex+1]["nama"]:
                data_list[jindex], data_list[jindex+1] = data_list[jindex+1], data_list[jindex]
    return data_list

def bubble_sort_by_harga(data_list):
    pjglist = len(data_list)
    for index in range(0, pjglist - 1):
        for jindex in range(0, pjglist - index - 1):
            if data_list[jindex]["harga"] > data_list[jindex+1]["harga"]:
                data_list[jindex], data_list[jindex+1] = data_list[jindex+1], data_list[jindex]
    return data_list

# GUI components

label = tk.Label(window, text = ("Welcome To Toko Jam  \n ⌚"),bg = "lightblue", foreground="darkblue", font = ("Helvetica", 20))
label.pack(pady = 50)

label = tk.Label(window, text = ("Please input your choice:"),bg = "lightblue", foreground="darkblue", font = ("Helvetica", 15))
label.pack(pady = 5)

label_options = tk.Label(window, text = ("1. Sort by Name \n2. Sort by Price \n3. Show Items \n4. Exit"),bg = "lightblue", justify ="center", foreground="darkblue", font = ("Helvetica", 12))
label_options.pack(pady = 5)

label_output = tk.Label(window, text = "",bg = "lightblue", fg = "orange", justify= "left", font = ("Helvetica", 12))
label_output.pack(pady = 5)
   
entry = tk.Entry (window, width = 30)
entry. pack(pady = 5)


# function to handle button press
def press():
    
    pilihan = entry.get()
    if pilihan == "1": 
        hasil = bubble_sort_by_nama(etalase)
        tampilkan_etalase (hasil)
    elif pilihan == "2":
        hasil = bubble_sort_by_harga(etalase)
        tampilkan_etalase(hasil)
    elif pilihan == "3":
        tampilkan_etalase(etalase)
    elif pilihan == "4":
        label_output.config(text="Thank you for shopping!")
      
    else: label_output.config(text="Please input a valid option")
button = tk.Button(window, text="See Choice", command = press)
button.pack(pady = 5)


entry_buy= tk.Entry (window, bg="lightblue", width = 30)
entry. pack(pady = 5)

# function to handle buy button press
def buy():
    item_name = entry_buy.get()
    for item in etalase:
        if item_name.lower()== item["nama"].lower():
            label_output.config(text=f"You have bought {item['nama']} for {item['harga']}. Thank you!")
            return
    label_output.config(text=f"You have bought {item['nama']} for {item['harga']}. Thank you!")
button_buy = tk.Button(window, text="Buy Item", command=buy)
button_buy.pack(pady = 5)

    
window.mainloop()
```
