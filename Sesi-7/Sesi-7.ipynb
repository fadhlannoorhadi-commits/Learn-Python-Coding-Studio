```python
# Kita import library
import tkinter as tk #library, as = alias untuk mempersingkat tkinter

# Kita membuat Window
window = tk.Tk()

# membuat judul dari window yang dibuat
window.title("my application")

# atur ukuran dair window
window.geometry ("300x250") #ini dalam bentuk px

# membuat label
label = tk.Label(window, text = "kelas python coding studio", font = ("calibry", 9))
label.pack(pady = 10) #tumpuk (ditumpuk diatas window biar keliatan)

label_2 = tk.Label(window, text = ("ini test aplikasi 1"))
label_2.pack(pady = 10) #pady = jarak dari sumbu y (untukl membuat jarak)

# jalankan app
window.mainloop()



```

```python
# Button
import tkinter as tk #import library

# membuat jendela
window = tk.Tk()

# judul dari window
window.title("My Application")

# atur besar
window.geometry("300x250")

# function
def klik():
    print("button diklik user....")

# label
label = tk.Label (window, text = "belajar membuat button", font = ("calibry", 15))
label.pack(pady=5)

# buat button
button = tk.Button(window, text = "klik saya", command = klik)
button.pack()

window.mainloop()
```

```python
# Manajemen Layout
# Pack (vertical/horizonta)
# Grid (berdasarkan koordinat)


# pack
import tkinter as tk
window = tk.Tk()
window.title("my application")

window.geometry("400x300")

tk.Label(window, text = "label 1") .pack()
tk.Button (window, text = "button 1") .pack()
tk.Label(window, text = "label 3") .pack()

window.mainloop()
```

```python
# Grid
import tkinter as tk
window = tk.Tk()
window.title("my application")
window.geometry("400x300")

# label
nama = tk.Label(window, text = "nama:")
nama.grid(row = 0, column = 0, pady=10,padx=10)

# buat entry nama
entry_nama = tk.Entry(window, width = 20)
entry_nama.grid(row = 0, column = 1, pady=10,padx=10)

# buat label entry email
email = tk.Label(window, text = "email:")
email.grid (row=1, column=0, pady=10,padx=10)

# buat entry email

entry_email=tk.Entry(window, width = 20)
entry_email.grid(row=1, column=1, pady=10,padx=10)

window.mainloop()
```

```python
# placer
import tkinter as tk
window = tk.Tk()
window.title("my application")
window.geometry("400x300")

button_1 = tk.Button(window, text = "button di posisi (x=30, y=50)")
button_1.place(x=30, y=50)

button_2 = tk.Button(window, text= "button di posisi (x=150, y=120)")
button_2.place(x=150, y=120)

window.mainloop()
```
